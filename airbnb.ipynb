```python
#There is host data present within the listings table that can be seperated into its own table.
host_cols = ['host_id',
       'host_url', 'host_profile_id', 'host_profile_url', 'host_name',
       'host_since', 'hosts_time_as_user_years', 'hosts_time_as_user_months',
       'hosts_time_as_host_years', 'hosts_time_as_host_months',
       'host_location', 'host_about', 'host_response_time',
       'host_response_rate', 'host_acceptance_rate', 'host_is_superhost',
       'host_thumbnail_url', 'host_picture_url', 'host_neighbourhood',
       'host_listings_count', 'host_total_listings_count',
       'host_verifications', 'host_has_profile_pic', 'host_identity_verified', 'calculated_host_listings_count',
       'calculated_host_listings_count_entire_homes',
       'calculated_host_listings_count_private_rooms',
       'calculated_host_listings_count_shared_rooms']
df_host = df[host_cols]

#This relationship from listings to hosts is many to one. The host data appears many times within the listings table, meaning it currently appears many times in the host table.
df_host = df_host.drop_duplicates()

#There are still duplicate ids in the table are removing duplicate rows
df_host["host_id"].value_counts()
```

```python
#It appears that remaining duplicate ids are due to host_profile_url being present in one instance but not the other.
duplicate_hosts = df_host["host_id"].value_counts()[df_host["host_id"].value_counts() > 1]
duplicate_ids = duplicate_hosts.index

df_host[df_host["host_id"].isin(duplicate_ids)].sort_values(by='host_id')
```

```python
#Since duplicate rows have are identical except for the profile url, we can drop the duplicate row without the profile url. This results in a table free of duplicate ids.
drop_idx = df_host[df_host["host_id"].isin(duplicate_ids) & df_host["host_profile_url"].isna()].index
df_host = df_host.drop(index = drop_idx)

len(df_host["host_id"].value_counts()[df_host["host_id"].value_counts() > 1])
```

```python
#these columns are dropped from the listings table, but host_id is kept in order to maintain the relationship
df_clean = df.drop([col for col in host_cols if col != "host_id"], axis=1)
df_clean.columns
```

```python
review_statistics_cols = ['id', 'number_of_reviews', 'number_of_reviews_ltm', 'number_of_reviews_l30d',
        'number_of_reviews_ly','first_review', 'last_review',
       'review_scores_rating', 'review_scores_accuracy',
       'review_scores_cleanliness', 'review_scores_checkin',
       'review_scores_communication', 'review_scores_location',
       'review_scores_value', 'reviews_per_month']
df_review_statistics = df_clean[review_statistics_cols]
df_review_statistics.rename(columns={'id': 'listing_id'}, inplace=True)

#This relationship from listings to hosts is one to one. There should not be any duplicates in this table since it is one for each listing
df_review_statistics["listing_id"].value_counts()
```
