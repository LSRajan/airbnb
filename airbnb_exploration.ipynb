##1.2 Host time as host

```python
df_host['hosts_time_as_host_years'] = df_host['hosts_time_as_host_years'].astype('Int64')
df_host['hosts_time_as_host_months'] = df_host['hosts_time_as_host_months'].astype('Int64')
```

```python
df_host["hosts_time_as_host_years"].describe()
```

```python
df_host['hosts_time_as_host_total'] = df_host['hosts_time_as_host_years']*12 + df_host['hosts_time_as_host_months']
df_host['hosts_time_as_host_total'].describe()
```

It appears the majority of hosts have been hosts for between 75-130 months (6.25-10.8 Years Ago). There is also a second peak in the graph at around 30 months (2.5 Years Ago). If we observe the gap between the two clusters, which is between 50-70 months (4.1-5.8 Years Ago) , we can see that there was likely a dip in new hosts around the time of COVID restrictions.

```python
plt.hist(df_host['hosts_time_as_host_total'], bins = 25)
plt.title("New Host Trends (from sample)")
plt.xlabel("Months Ago")
plt.ylabel("Number of New Hosts")
```

The last scraped value allows us to see the reference point for the 'months ago' value. The values align with the hypothesis that the dip in new hosts is due to COVID restrictions.

```python
df_listings['last_scraped'].value_counts()
```

```python
df_host['hosts_time_as_user_total'] = df_host['hosts_time_as_user_years']*12 + df_host['hosts_time_as_user_months']
df_host['hosts_time_as_user_total'].describe()
```

Interestingly most of the hosts became users during the same period, despite the more recent surge in users becoming hosts. The graphs show a steady intake of new hosts, with the exception of COVID, but a majority of new users from one time period. This could suggest that most users created accounts during a surge of interest in airbnb, and said users progressively became hosts.

```python
plt.hist(df_host['hosts_time_as_user_total'], bins = 25)
plt.title("New User Trends (Only future hosts from sample)")
plt.xlabel("Months Ago")
plt.ylabel("Number of New Users")
plt.axvline(x=125, color='grey', linestyle="--", label='Median')
plt.legend()
```

```python
df_host['hosts_time_until_host'] = df_host['hosts_time_as_user_total'] - df_host['hosts_time_as_host_total']
df_host['hosts_time_until_host'].describe()
```

The majority of hosts became hosts shortly after becoming users, suggesting most hosts created accounts with the intention of becoming hosts. On the other hand, 25% of users became hosts after 45 months (over 2 Years) of being a user.

```python
plt.hist(df_host['hosts_time_until_host'], bins = 25)
plt.axvline(x=45, color='grey', linestyle="--", label='75 percentile')
plt.legend()
```

##1.6 Time as Host relationships

To inspect the relationship between the length of time a host has been hosting, we first need to create bins to create categories to compare between.

```python
print(df_host["hosts_time_as_host_total"].min())
print(df_host["hosts_time_as_host_total"].max())
```

```python
df_host['hosts_time_as_host_total'] = df_host['hosts_time_as_host_total']/12
df_host['hosts_time_as_user_total'] = df_host['hosts_time_as_user_total']/12


bins = np.arange(0, 200/12)
df_host['hosts_time_as_host_binned'] = pd.cut(df_host['hosts_time_as_host_total'], bins)
df_host['hosts_time_as_host_binned']
```

By observing the mean time a host spent as a user across bins, we see that for hosts who have been hosts for much longer became hosts shortly after becoming a user. In general the longer a host has been a host, the less time they had been a user for.

```python
df_host.groupby('hosts_time_as_host_binned')['hosts_time_as_user_total'].mean().round(2)
```

```python
df_host.groupby(df_host["host_location"].isna())['hosts_time_as_host_total'].mean().round(2)
```

It appears that hosts who are located outside of Germany are on average much younger accounts, (about 4 years younger). They have also been on average about hosts for about 3 years less. This likely points to a more recent shift towards commercial properties.

```python
df_host.groupby(df_host['host_country'] == "Germany").agg({
    "hosts_time_as_host_total" : ["mean", "count"],
    "hosts_time_as_user_total" : ["mean", "count"]
})
```

Hosts tend to have about sections that are similar in lengths across bins, however hosts who hen hosting for longer are on average more likely to have an about section written at all. Infact hosts who have become hosts in the last 6 years each have less than 40% of their bin with about sections written.

```python
df_host.groupby(df_host['hosts_time_as_host_binned']).agg({
    "host_about" : [("average_length", lambda x: round(x.dropna().str.len().mean(), 2)),
                     ("pct_not_missing", lambda x: round(x.count() / x.size * 100, 2)),
                    "count"]
})
```

The data suggests that being a superhost is unrelated to a hosts time spent as a host.

```python
df_host.groupby(df_host['host_is_superhost']).agg({
    "hosts_time_as_host_total" : ["mean", "count"],
    "hosts_time_as_user_total" : ["mean", "count"]
})
```

It seems there are two peaks in the table. Around 25-30% of users who have been host for 1-5 years are superhosts. Then at least 30% of users who have been hosts for longer than 13 years are superhosts, with 50% of the 58 hosts who have been hosts for 15-16 years are superhosts. Remaining bins see around 20% of users as superhosts.

```python
df_host.groupby(df_host['hosts_time_as_host_binned']).agg({
    "host_is_superhost" : [("pct_superhost", lambda x: round(x.sum() / x.size * 100, 2)),
                     ("pct_not_missing", lambda x: round(x.count() / x.size * 100, 2)),
                           "count"]
})
```

```python
df_host.groupby(df_host['host_picture_url'].isna()).agg({
    "hosts_time_as_host_total" : ["mean", "count"],
    "hosts_time_as_user_total" : ["mean", "count"]
})
```

As we have seen with other variables, users with missing information tend to have been hosts for much less time on average.

```python
df_host.groupby(df_host['host_has_profile_pic']).agg({
    "hosts_time_as_host_total" : ["mean", "count"],
    "hosts_time_as_user_total" : ["mean", "count"]
})
```

```python
df_host.groupby(df_host['hosts_time_as_host_binned']).agg({
    "host_has_profile_pic" : [("pct_has_profile_pic", lambda x: round(x.sum() / x.size * 100, 2)),
                     ("pct_not_missing", lambda x: round(x.count() / x.size * 100, 2)),
                              "count"]
})
```

```python
df_host.groupby(df_host['host_identity_verified']).agg({
    "hosts_time_as_host_total" : ["mean", "count"],
    "hosts_time_as_user_total" : ["mean", "count"]
})
```

```python
df_host.groupby(df_host['hosts_time_as_host_binned']).agg({
    "host_identity_verified" : [("pct_verified", lambda x: round(x.sum() / x.size * 100, 2)),
                     ("pct_not_missing", lambda x: round(x.count() / x.size * 100, 2)),
                                "count"]
})
```

```python
df_host.groupby(df_host['license'].isna()).agg({
    "hosts_time_as_host_total" : ["mean", "count"],
    "hosts_time_as_user_total" : ["mean", "count"]
})
```

Interestingly, the presence of a license contradicts the pattern of newer hosts having missing data. Since hosts who have been hosting 5 years or less have around 75-80%, whereas hosts who have been hosting between 5 and 13 years have around 50-60% on average

```python
df_host.groupby(df_host['hosts_time_as_host_binned']).agg({
    "license" : [("pct_not_missing", lambda x: round(x.count() / x.size * 100, 2)),
                 "count"]
})
```

```python
df_host.groupby(df_host['hosts_time_as_host_binned']).agg({
    'calculated_host_listings_count': ["mean", "median",
                     ("pct_not_missing", lambda x: round(x.count() / x.size * 100, 2)),
                                       "count"],
    'calculated_host_listings_count_entire_homes': ["mean", "median",
                     ("pct_not_missing", lambda x: round(x.count() / x.size * 100, 2)),
                                       "count"],
    'calculated_host_listings_count_private_rooms': ["mean", "median",
                     ("pct_not_missing", lambda x: round(x.count() / x.size * 100, 2)),
                                       "count"],
    'calculated_host_listings_count_shared_rooms': ["mean", "median",
                     ("pct_not_missing", lambda x: round(x.count() / x.size * 100, 2)),
                                       "count"],
})
```
