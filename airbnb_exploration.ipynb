##1.2 Host time as host

```python
df_host['hosts_time_as_host_years'] = df_host['hosts_time_as_host_years'].astype('Int64')
df_host['hosts_time_as_host_months'] = df_host['hosts_time_as_host_months'].astype('Int64')
```

```python
df_host["hosts_time_as_host_years"].describe()
```

```python
df_host['hosts_time_as_host_total'] = df_host['hosts_time_as_host_years']*12 + df_host['hosts_time_as_host_months']
df_host['hosts_time_as_host_total'].describe()
```

It appears the majority of hosts have been hosts for between 75-130 months (6.25-10.8 Years Ago). There is also a second peak in the graph at around 30 months (2.5 Years Ago). If we observe the gap between the two clusters, which is between 50-70 months (4.1-5.8 Years Ago) , we can see that there was likely a dip in new hosts around the time of COVID restrictions.

```python
plt.hist(df_host['hosts_time_as_host_total'], bins = 25)
plt.title("New Host Trends (from sample)")
plt.xlabel("Months Ago")
plt.ylabel("Number of New Hosts")
```

The last scraped value allows us to see the reference point for the 'months ago' value. The values align with the hypothesis that the dip in new hosts is due to COVID restrictions.

```python
df_listings['last_scraped'].value_counts()
```

```python
df_host['hosts_time_as_user_total'] = df_host['hosts_time_as_user_years']*12 + df_host['hosts_time_as_user_months']
df_host['hosts_time_as_user_total'].describe()
```

Interestingly most of the hosts became users during the same period, despite the more recent surge in users becoming hosts. The graphs show a steady intake of new hosts, with the exception of COVID, but a majority of new users from one time period. This could suggest that most users created accounts during a surge of interest in airbnb, and said users progressively became hosts.

```python
plt.hist(df_host['hosts_time_as_user_total'], bins = 25)
plt.title("New User Trends (Only future hosts from sample)")
plt.xlabel("Months Ago")
plt.ylabel("Number of New Users")
plt.axvline(x=125, color='grey', linestyle="--", label='Median')
plt.legend()
```

```python
df_host['hosts_time_until_host'] = df_host['hosts_time_as_user_total'] - df_host['hosts_time_as_host_total']
df_host['hosts_time_until_host'].describe()
```

The majority of hosts became hosts shortly after becoming users, suggesting most hosts created accounts with the intention of becoming hosts. On the other hand, 25% of users became hosts after 45 months (over 2 Years) of being a user.

```python
plt.hist(df_host['hosts_time_until_host'], bins = 25)
plt.axvline(x=45, color='grey', linestyle="--", label='75 percentile')
plt.legend()
```

##1.3 Host location



```python
df_host["host_location"].value_counts(dropna=False)
```

```python
split = df_host["host_location"].str.split(", ", n=1, expand=True)

df_host["host_city"] = split[0]
df_host["host_country"] = split[1]


#For values such as "Germany", they will return ["Germany", NA], but we need [NA, "Germany"]
singular_values = df_host["host_city"].notna() & df_host["host_country"].isna()
df_host.loc[singular_values, ["host_city", "host_country"]]
```

```python
df_host.loc[singular_values,"host_country"] = df_host.loc[singular_values,"host_city"]
df_host.loc[singular_values,"host_city"] = pd.NA
```

```python
df_host["host_country"].value_counts()
```

```python
df_host["host_city"].value_counts()
```

```python
top_countries = df_host["host_country"].value_counts().head(11)
top_cities = df_host["host_city"].value_counts().head(10)
```

```python
top_countries
```

Here we can see that Germany and Berlin represent the majority for host locations. This is to be espected since the data is for airbnbs in Berlin, Germany

```python
sorted_countries = top_countries.sort_values(ascending=True)

plt.barh(sorted_countries.index, sorted_countries.values)
```

```python
sorted_cities = top_cities.sort_values(ascending=True)

plt.barh(sorted_cities.index, sorted_cities.values)
```

Looking more closely at the rest of the locations, it seems that the remaining hosts are spread across surrounding countries and a few notable countiries/states. This does not produce any unsurprising insights but does suggest the data follows typical global patterns.

```python
sorted_countries = top_countries[1:].sort_values(ascending=True)

plt.barh(sorted_countries.index, sorted_countries.values)
```

```python
sorted_cities = top_cities[1:].sort_values(ascending=True)

plt.barh(sorted_cities.index, sorted_cities.values)
```
