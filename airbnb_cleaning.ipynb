# 1.2 Host Table Cleaning --------------

The listings table has host that can be seperated into its own table.
Since the relationship from listings to hosts is many to one, the host data appears many times within the listings table.
 However, after removing these duplicate rows, we find there are still duplicate ids in the table.

```python
host_cols = ['host_id',
       'host_url', 'host_profile_id', 'host_profile_url', 'host_name', 'hosts_time_as_user_years', 'hosts_time_as_user_months',
       'hosts_time_as_host_years', 'hosts_time_as_host_months',
       'host_location', 'host_about', 'host_is_superhost', 'host_picture_url',
       'host_listings_count',
      'host_has_profile_pic', 'host_identity_verified', 'calculated_host_listings_count',
       'calculated_host_listings_count_entire_homes',
       'calculated_host_listings_count_private_rooms',
       'calculated_host_listings_count_shared_rooms']
df_host = df[host_cols]

df_host = df_host.drop_duplicates()
df_host["host_id"].value_counts()
```

Upon further inspection, it appears that remaining duplicate ids are due to  values for the 'host_profile_url' column being present in one instance but not the other.

```python
duplicate_hosts = df_host["host_id"].value_counts()[df_host["host_id"].value_counts() > 1]
duplicate_ids = duplicate_hosts.index

df_host[df_host["host_id"].isin(duplicate_ids)].sort_values(by='host_id')
```

Since duplicate rows have are identical except for the profile url, we can drop the duplicate row without the profile url. This results in a table free of duplicate ids.

```python
drop_idx = df_host[df_host["host_id"].isin(duplicate_ids) & df_host["host_profile_url"].isna()].index
df_host = df_host.drop(index = drop_idx)

len(df_host["host_id"].value_counts()[df_host["host_id"].value_counts() > 1])
```

The host columns can be dropped from the listings table, but host_id is kept in order to maintain the relationship

```python
df_clean = df.drop([col for col in host_cols if col != "host_id"], axis=1)
df_clean.columns
```

Now that the table structure is valid, we can work on cleaning the remaining columns

```python
df_host.head()
```

The host profile url is redundant since it can be generated directly from the host id. Thus we will remove this column to simplify the overall table.

```python
print(len(df_host[df_host["host_url"] != "https://www.airbnb.com/users/show/" + df_host["host_id"].astype(str)]))
df_host = df_host.drop('host_url', axis=1)
```

We can apply similar logic to remove the 'host_profile_url' column. These columns are the same as far as is output, despite checks returning differences. Regardless, the url column is unlikely to be useful for exploration. Therefore, this column will be dropped too.

```python
df_host['host_profile_id'] = df_host['host_profile_id'].astype('Int64')

print("Number of differing rows: "  + str(len(
    df_host[
        (df_host["host_profile_url"] != "https://www.airbnb.com/users/profile/" + df_host["host_profile_id"].astype(str)) &
        df_host["host_profile_id"].notna()
        ])))

df_host.loc[df_host["host_profile_url"] != ("https://www.airbnb.com/users/profile/" + df_host["host_profile_id"].astype(str)),
            ["host_profile_url", "host_profile_id"]]
df_host = df_host.drop('host_profile_url', axis=1)
```

The name column has very few missing values and appears to be free of errors

```python
print("Number of NA values in name column: " + str(df_host["host_name"].isna().sum()))
print("Number of empty values in name column: " + str((df_host["host_name"] == "").sum()))
df_host["host_name"].head(20)
```

We can check the logic of the hosts_time_as_... columns. Since a host has to have been a user at least as long as they have been a host, we can check all rows for this condition. We can also ensure that months are between the intended range.

```python
time_as_cols = ['hosts_time_as_user_years',	'hosts_time_as_user_months',	'hosts_time_as_host_years',	'hosts_time_as_host_months']
df_host[time_as_cols] = df_host[time_as_cols].apply(lambda col: col.astype('Int64')) # converting unnecessary float cols to ints

print("Number of rows with host years > user years: " + str(len(df_host[df_host['hosts_time_as_user_years'] < df_host['hosts_time_as_host_years']])))
print(df_host.loc[df_host['hosts_time_as_user_years'] < df_host['hosts_time_as_host_years'],
                      ["hosts_time_as_user_years", "hosts_time_as_host_years"]])

print("Number of rows with host years + month > user years + month: " + str(len(df_host[
    (df_host['hosts_time_as_user_years'] == df_host['hosts_time_as_host_years']) &
    (df_host['hosts_time_as_user_months'] < df_host['hosts_time_as_host_months'])
    ])))
print(df_host.loc[(df_host['hosts_time_as_user_years'] == df_host['hosts_time_as_host_years']) &
    (df_host['hosts_time_as_user_months'] < df_host['hosts_time_as_host_months']),
                      ["hosts_time_as_user_years", "hosts_time_as_host_years"]])

#months should be between 0 and 12
print("Number of rows with user months >= 12 or < 0: " + str(len(df_host[
    (df_host['hosts_time_as_user_months'] >= 12) | (df_host['hosts_time_as_user_months'] < 0)
    ])))
print(df_host.loc[df_host['hosts_time_as_user_months'] >= 12,
                      ["hosts_time_as_user_years", "hosts_time_as_host_years"]])

print("Number of rows with host months >= 12 or < 0: " + str(len(df_host[
    (df_host['hosts_time_as_host_months'] >= 12) | (df_host['hosts_time_as_host_months'] < 0)
    ])))
print(df_host.loc[df_host['hosts_time_as_host_months'] >= 12,
                 ["hosts_time_as_user_years", "hosts_time_as_host_years"]])
```

The 4 rows with errors are dropped for consistency.

```python
drop_idx = df_host.loc[df_host['hosts_time_as_user_years'] < df_host['hosts_time_as_host_years'],
                      ["hosts_time_as_user_years", "hosts_time_as_host_years"]].index
df_host = df_host.drop(index = drop_idx)

drop_idx = df_host[
    (df_host['hosts_time_as_user_years'] == df_host['hosts_time_as_host_years']) &
    (df_host['hosts_time_as_user_months'] < df_host['hosts_time_as_host_months'])].index
df_host = df_host.drop(index = drop_idx)
```

```python
bool_cols = ['host_is_superhost',	'host_has_profile_pic',	'host_identity_verified']
df_host[bool_cols] = df_host[bool_cols].apply(lambda col: col.map({'t':True, 'f':False}))
df_host[bool_cols].head()
```

```python
count_cols = ['host_listings_count', 'calculated_host_listings_count',	'calculated_host_listings_count_entire_homes',	'calculated_host_listings_count_private_rooms',	'calculated_host_listings_count_shared_rooms']
for col in count_cols:
  print(df_host[col].value_counts(dropna=False).head())

df_host[count_cols] = df_host[count_cols].apply(lambda col: col.astype('Int64'))
```
